## Задание
1. Считать любое из тестовых изображений (в папке test_images)
2. Привести к полутоновому формату и провести бинаризацию
3. Применить морфологическую обработку для устранения шумов
4. Выполнить сегментацию объектов подходящим методом (например, методом водораздела)
5. Визуализировать метки сегментации на исходном изображении (например, выделить отдельные объекты разными цветами)

Например, можно использовать, как базовый, пример с монетами: **процесс сегментации, реализуемый в библиотеке OpenCV** (раздел 3.4)
